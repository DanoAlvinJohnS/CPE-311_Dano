```python
import numpy as np
import matplotlib.pyplot as plt


```

Given

```python
b = 1
w1 = 3
x1 = 0
```

Formula

```python
y = b + w1 * x1

print("Prediction for x1=0:", y)

x_vals = np.linspace(-10, 10, 50)
y_vals = b + w1 * x_vals
```

# Visualization

```python
plt.figure()
plt.plot(x_vals, y_vals, label="y = 1 + 3x")
plt.scatter(x1, y, label="Given Point (0,1)")
plt.xlabel("x1")
plt.ylabel("y")
plt.title("Single Feature Linear Regression")
plt.legend()
plt.grid()
plt.show()
```

#Analysis
The model is a straight line:
y=1+3x, Slope = 3 => y increases by 3 for every +1 increase in x,
Intercept = 1 => line crosses y-axis at (0,1). The point (0,1) lies exactly on the line, indicating that the prediction is solely influenced by the bias. This shows that even in the absence of input contribution, the model still produces a baseline output. The plotted graph confirms this relationship, as the point (0,1) lies exactly on the regression line, and the linear trend increases steadily due to the positive weight.

# Part 2

Given & Calculations

```python
b = 1
w = np.array([3, -2, 4])

x1 = np.array([3, -1, 2])
y1 = b + np.dot(w, x1)

x2 = np.array([10, 0, -10])
y2 = b + np.dot(w, x2)
print("Prediction for Example 1:", y1)
print("Prediction for Example 2:", y2)
```

Plotting

```python
x_vals = np.linspace(-10, 10, 50)
y_vals = 1 + 3*x_vals

plt.figure()
plt.plot(x_vals, y_vals, label="y = 1 + 3x1 (x2=0, x3=0)")
plt.scatter(3, 20, label="Example 1 (projected)")
plt.scatter(10, -9, label="Example 2 (projected)")
plt.xlabel("x1")
plt.ylabel("y")
plt.title("Projection of Multi-Feature Regression")
plt.legend()
plt.grid()
plt.show()
```

#Analysis
In the multi-feature scenario, the model becomes more complex as it incorporates three different variables, each contributing differently to the final prediction. The weights indicate both the strength and direction of influence 3 and 4 contribute positively, while −2 contributes negatively. For Additional Example #1, all feature contributions result in a large positive value, yielding a prediction of 20. This occurs because both positive weights and the negative weight (when multiplied by a negative input) add constructively to the output. In contrast, Additional Example #2 produces a negative prediction of -9, primarily due to the strong negative contribution from  −10 when multiplied by the large positive weight =4. This demonstrates how certain features can dominate the model’s output depending on their magnitude and associated weights.
